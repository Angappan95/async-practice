```python
# threads

import time
from concurrent.futures import ThreadPoolExecutor

def fetch_data(url):
    time.sleep(2)
    return f"Data fetched from {url}"

start = time.perf_counter()

with ThreadPoolExecutor(max_workers=3) as pool:
    # result = list(pool.map(fetch_data, ["url1", "url2", "url3"]))
    future1 = pool.submit(fetch_data, "url1")
    future2 = pool.submit(fetch_data, "url2")
    result = future1.result() + future2.result()


# print(result)
print(time.perf_counter()-start)
```

```python
# multi processing

from concurrent.futures import ProcessPoolExecutor

def calculate(n):
    return sum(i * i for i in range(n))

start = time.perf_counter()

with ProcessPoolExecutor(max_workers=50) as pool:
    result = list(pool.map(calculate, [100000]*100))

end = time.perf_counter()

# print(result)
print(end-start)
```

```python
import asyncio
import time

async def fetch_data(name):
    await asyncio.sleep(2)
    return f"{name} completed"

async def main():
    start = time.perf_counter()
    # result = await asyncio.gather(
    #     fetch_data("url1"),
    #     fetch_data("url2"),
    #     fetch_data("url3")
    # )
    urls = ["url1", "url2", "url3"]
    await asyncio.gather(*(fetch_data(url) for url in urls))
    end = time.perf_counter()
    # print(result)
    print(f"Elapsed time: {end-start:.2f} seconds")

# In Jupyter/Colab notebooks, run directly using top-level await:
await main()
```
